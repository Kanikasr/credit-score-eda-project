## 3. ***Data Wrangling***

### Data Wrangling Code

```python
# Write your code to make your dataset analysis ready.
# Step 0: Copy original DataFrame
df_cleaned = data.copy()

# Step 1: Drop identifier columns
columns_to_drop = ['ID', 'Customer_ID', 'Name', 'SSN']
df_cleaned.drop(columns=columns_to_drop, inplace=True)

# Step 2: Replace placeholders with np.nan
missing_values = ['No Data', 'NA', 'Unknown', 'N/A', '--', '', ' ','-']
df_cleaned.replace(missing_values, np.nan, inplace=True)

# Step 3: Standardize column names
df_cleaned.columns = (
    df_cleaned.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_')
    .str.replace('-', '_')
    .str.replace(r'[^0-9a-zA-Z_]', '', regex=True)
)

# Step 4: Convert categorical columns to category dtype
categorical_columns = [
    'occupation', 'type_of_loan', 'credit_mix',
    'payment_of_min_amount', 'payment_behaviour', 'credit_score'
]
df_cleaned[categorical_columns] = df_cleaned[categorical_columns].astype('category')

# Step 5: Convert integer-like columns
int_columns = [
    'num_bank_accounts', 'num_credit_card', 'interest_rate',
    'num_of_loan', 'delay_from_due_date', 'num_of_delayed_payment',
    'num_credit_inquiries', 'credit_history_age'
]
df_cleaned[int_columns] = df_cleaned[int_columns].round().astype('Int64')

# Step 6: Log-transform skewed numeric columns
skewed_cols = ['annual_income', 'outstanding_debt', 'amount_invested_monthly', 'monthly_balance']
for col in skewed_cols:
    df_cleaned[f'log_{col}'] = np.log1p(df_cleaned[col])

# Step 7: Clean 'type_of_loan' and create binary features
df_cleaned['type_of_loan'] = df_cleaned['type_of_loan'].astype(str).str.lower().fillna('')
df_cleaned['type_of_loan'] = df_cleaned['type_of_loan'].str.replace(r'\s+and\s+', ',', regex=True)
placeholders = ['not specified', 'no data', 'nan', 'unknown']
for ph in placeholders:
    df_cleaned['type_of_loan'] = df_cleaned['type_of_loan'].str.replace(ph, '', regex=False)
df_cleaned['type_of_loan'] = df_cleaned['type_of_loan'].str.replace(r',+', ',', regex=True)
df_cleaned['type_of_loan'] = df_cleaned['type_of_loan'].str.strip(',').str.strip()
df_cleaned['type_of_loan_list'] = df_cleaned['type_of_loan'].apply(
    lambda x: list(set([loan.strip().title() for loan in x.split(',') if loan.strip()]))
)
unique_loans = set(loan for sublist in df_cleaned['type_of_loan_list'] for loan in sublist)
for loan in unique_loans:
    df_cleaned[f'loan_{loan.lower().replace(" ", "_")}'] = df_cleaned['type_of_loan_list'].apply(
        lambda x: 1 if loan in x else 0
    )
df_cleaned.drop(columns=['type_of_loan_list', 'type_of_loan'], inplace=True)

# Step 8: Convert 'payment_of_min_amount' to binary
df_cleaned['payment_of_min_amount'] = df_cleaned['payment_of_min_amount'].astype(str).str.strip().str.lower()
df_cleaned['payment_min_binary'] = df_cleaned['payment_of_min_amount'].map({
    'yes': 1,
    'no': 0,
    'nm': np.nan
})

# Step 9: Detect and flag outliers
numeric_cols = [
    'annual_income', 'monthly_balance', 'outstanding_debt',
    'credit_utilization_ratio', 'amount_invested_monthly', 'total_emi_per_month'
]

def detect_outliers_iqr(df, col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return (df[col] < lower_bound) | (df[col] > upper_bound)

for col in numeric_cols:
    df_cleaned[f'{col}_outlier'] = detect_outliers_iqr(df_cleaned, col)

# Step 10: Outlier treatment
# (1) Log-transform for skewed columns (again for safety)
log_transform_cols = ['annual_income', 'outstanding_debt', 'monthly_balance']
for col in log_transform_cols:
    df_cleaned[f'log_{col}'] = np.log1p(df_cleaned[col])

# (2) Capping (Winsorization)
def cap_outliers(df, col, lower_percentile=0.01, upper_percentile=0.99):
    lower_bound = df[col].quantile(lower_percentile)
    upper_bound = df[col].quantile(upper_percentile)
    df[col] = np.where(df[col] < lower_bound, lower_bound, df[col])
    df[col] = np.where(df[col] > upper_bound, upper_bound, df[col])
    return df

cap_only_cols = ['num_credit_inquiries', 'delay_from_due_date', 'changed_credit_limit']
for col in cap_only_cols:
    df_cleaned = cap_outliers(df_cleaned, col)

# (3) Combination: capping + log for important columns
cap_and_log_cols = ['total_emi_per_month', 'amount_invested_monthly']
for col in cap_and_log_cols:
    df_cleaned = cap_outliers(df_cleaned, col)
    df_cleaned[f'log_{col}'] = np.log1p(df_cleaned[col])

# Final column name formatting
df_cleaned.columns = df_cleaned.columns.str.replace('-', '_')

# Save cleaned data
# df_cleaned.to_csv('cleaned_credit_data.csv', index=False)
```

### What all manipulations have you done and insights you found?

To begin the data preparation process, we created a copy of the original dataset to ensure that our cleaning operations did not alter the raw data. We then removed identifier columns such as **ID, Customer_ID, Name, and SSN,** as these fields do not contribute any analytical or predictive value and may pose privacy concerns. Next, we standardized missing values by replacing common placeholders like **'No Data', 'NA', 'Unknown', and blanks with np.nan**, enabling consistent handling of missing entries. We also standardized column names by converting them to **lowercase, replacing spaces and special characters** with underscores to make the dataset cleaner and more manageable in code.

To optimize memory usage and prepare for categorical analysis, we explicitly converted categorical columns such as **occupation, type_of_loan, credit_mix, payment_of_min_amount, payment_behaviour, and credit_score to the category data** type. Numeric columns that represented counts or discrete values, such as the **number of credit cards or credit inquiries, were rounded and converted to integer format** to better reflect their actual meaning and to eliminate unnecessary decimal values.

Since several **numeric columns were highly skewed (e.g., annual_income, outstanding_debt, amount_invested_monthly, and monthly_balance), we applied logarithmic transformation (log1p)** to reduce skewness and stabilize variance, making them more suitable for visualizations and modeling. For the **type_of_loan column, which included multiple loan types in a single string, we cleaned and split the values into individual categories, creating binary columns** for each loan type to enable multi-label analysis.

The **payment_of_min_amount column**, which originally contained categorical responses such as 'Yes', 'No', and 'NM', was mapped to a binary format to simplify interpretation. We also detected outliers in key **numeric columns using the interquartile range (IQR) method and flagged them** accordingly. For certain columns with extreme values, such as **num_credit_inquiries and delay_from_due_date, we applied capping (Winsorization)** to limit their range without removing data. For **heavily skewed and important columns like total_emi_per_month and amount_invested_monthly, we used a combination of capping and log transformation** to bring them within a manageable range.

Finally, we saved the fully cleaned and transformed dataset to a new CSV file (cleaned_credit_data.csv) for use in subsequent exploratory data analysis and modeling. This comprehensive preprocessing ensured the data was clean, consistent, and analytically robust.

#### Chart - 7 Credit Score vs Avg utilization (Bar plot),  Category (credit_score) vs Numeric (avg utilization)

#### Chart - 9 Credit Score vs Occupation(Stacked Bar Plot), Category vs Category

```python
# Chart - 9 visualization code
# Group by occupation and credit_score to count records
occupation_score = df.groupby(['occupation', 'credit_score']).size().reset_index(name='Count')

# Pivot to wide format for stacking
occupation_score_pivot = occupation_score.pivot(index='occupation', columns='credit_score', values='Count').fillna(0)

# Reorder columns for visual consistency
credit_order = ['Standard', 'Poor', 'Good']
occupation_score_pivot = occupation_score_pivot[credit_order]

# Normalize rows to get percentage distribution
occupation_score_pct = occupation_score_pivot.div(occupation_score_pivot.sum(axis=1), axis=0)

# Sort occupations by proportion of Good credit score
occupation_score_pct = occupation_score_pct.sort_values(by='Good', ascending=False)

# Plot
ax = occupation_score_pct.plot(
    kind='bar', stacked=True, figsize=(14, 7), colormap='Set2', legend=True
)
plt.title('Proportion of Credit Scores by Occupation')
plt.ylabel('Proportion (0 to 1)')
plt.xlabel('Occupation')
plt.xticks(rotation=45)
plt.legend(title='Credit Score')

# Annotate percentage labels on bars
for i, col in enumerate(occupation_score_pct.columns):
    for j, (idx, val) in enumerate(occupation_score_pct[col].items()):
        cumulative = occupation_score_pct.iloc[j, :i].sum()
        if val > 0.03:  # Avoid clutter for small segments
            ax.text(j, cumulative + val / 2, f"{val:.2f}", ha='center', va='center', fontsize=8)

plt.tight_layout()
plt.show()
```

This chart helps visualize how credit scores are distributed across different occupations, enabling a clear comparison of financial behavior among job groups.

#### Chart - 21 Heatmap of Loan Type vs Credit Score (Normalized %), Category vs Category (normalized frequency)


```python

# Step 1: Select loan columns
loan_cols = [col for col in df.columns if col.startswith('loan_')]

# Step 2: Melt the DataFrame into long format
loan_data = df.melt(id_vars='credit_score', value_vars=loan_cols,
                    var_name='loan_type', value_name='has_loan')

# Step 3: Keep only rows where a loan exists
loan_data = loan_data[loan_data['has_loan'] == 1]

# Step 4: Clean the loan_type names
loan_data['loan_type'] = loan_data['loan_type'].str.replace('loan_', '').str.replace('_', ' ').str.title()

# Step 5: Count by loan type and credit score
grouped = loan_data.groupby(['loan_type', 'credit_score']).size().reset_index(name='count')

# Step 6: Pivot and normalize row-wise (%)
pivot = grouped.pivot(index='loan_type', columns='credit_score', values='count').fillna(0)
pivot_percent = pivot.div(pivot.sum(axis=1), axis=0) * 100  # normalize rows to percentages

# Step 7: Plot the heatmap
plt.figure(figsize=(12,7))
sns.heatmap(pivot_percent, annot=True, fmt=".1f", cmap='YlGnBu', linewidths=0.5, linecolor='gray')

plt.title('Normalized % of Credit Score by Loan Type')
plt.xlabel('Credit Score')
plt.ylabel('Loan Type')
plt.tight_layout()
plt.show()

```

##### 1. Why did you pick the specific chart?

Heatmaps reveal pattern intensity across combinations.

Normalizing by row (loan type) allows fair comparison even when raw loan counts differ.

Useful to spot which credit score is overrepresented or underrepresented for each loan type.



##### 2. What is/are the insight(s) found from the chart?

Standard credit scores dominate all loan types (~50% share).

Poor scores are consistently high (≈37–38%) across loans — especially Auto and Student loans.

Good scores are low across all loans (<13%), with Student loans having the lowest good-score ratio (12.5%).
